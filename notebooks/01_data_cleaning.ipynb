```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/Airbnb_Open_Data.csv", engine="python")
df.head()
```

```python
df.shape
df.columns
df.info()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df.columns
```

```python
df = df.drop(columns=["license", "house_rules"])
```

```python
df["price"] = df["price"].replace(r"[\$,]", "", regex=True)
df["service_fee"] = df["service_fee"].replace(r"[\$,]", "", regex=True)

df["price"] = pd.to_numeric(df["price"], errors="coerce")
df["service_fee"] = pd.to_numeric(df["service_fee"], errors="coerce")
```

```python
df["instant_bookable"] = df["instant_bookable"].astype(str).str.lower().str.strip()
df["instant_bookable"] = df["instant_bookable"].replace("nan", pd.NA)
df = df.dropna(subset=["instant_bookable"])
```

```python
df["ab_group"] = df["instant_bookable"].apply(
    lambda x: "treatment" if x == "true" else "control"
)
```

```python
df["reviews_per_month"] = df["reviews_per_month"].fillna(0)
```

```python
df.to_csv("../data/processed/airbnb_cleaned.csv", index=False)
```
